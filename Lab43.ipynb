```python
# Lab 4-8 Prime numbers
# Pseduocode: to decide N is prime or not
# for number in 2 to N -1
#   if N is divisible by number;
#      N is not prime;
#      break;

# Exist x, x > 10 , x = {12, 3, 5, 7, 15}
# You can decide True in the middle of the set, but you can not decide True in the end of the set.
# You can decide False at the end of the set.

# Not exist x, x > 10 , x = {12, 3, 5, 7, 15}
# You can false in the middle of the set, but you can not false in the end of the set.
```

```python
# a1 = int(input("Enter a number: "))
# a2 = int(input("Enter another number: "))
# N = int(input("Enter a number: "))
result = []
a1 = 2
a2 = 3
N = 10
# print(a1, a2, end=' ')
result.append(a1); result.append(a2)
for i in range(N - 2):
    next = a1 + a2
    a1 = a2
    a2 = next
    # print(next, end=' ')
    result.append(next)
print(result)
```
